```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,f1_score
from sklearn.model_selection import KFold,cross_val_score
```

```python
iris=load_iris()
x=iris.data
y=iris.target
print("Dataset shape:",x.shape)
```

```python
#Bootstrapping
np.random.seed(42)
n_iterations=10
bootstrap_accuracy=[]
bootstrap_f1=[]
for i in range(n_iterations):
    indices=np.random.choice(
        len(x),
        size=len(x),
        replace=True
    )
    x_train=x[indices]
    y_train=y[indices]
    test_indices=list(set(range(len(x)))-set(indices))
    x_test=x[test_indices]
    y_test=y[test_indices]
    model=DecisionTreeClassifier(random_state=42)
    model.fit(x_train,y_train)
    pred=model.predict(x_test)
    accuracy=accuracy_score(y_test,pred)
    f1=f1_score(y_test,pred,average="weighted")
    bootstrap_accuracy.append(accuracy)
    bootstrap_f1.append(f1)
print("\nBootstrapping Results")
print("-----------------------")
print("Average Accuracy:",round(np.mean(bootstrap_accuracy),3))
print("Average F1 Score:",round(np.mean(bootstrap_f1),3))
```

```python
#K Fold
kf=KFold(n_splits=5,
         shuffle=True,
         random_state=42
        )
model=DecisionTreeClassifier(random_state=42)
accuracy_scores=cross_val_score(model,x,y,cv=kf,scoring="accuracy")
f1_scores=cross_val_score(model,x,y,cv=kf,scoring="f1_weighted")
#results
print("\n5-Fold Coss-Validation Results")
print("------------------------")
print("Average Accuracy:",round(np.mean(accuracy_scores),3))
print("Average F1 Score:",round(np.mean(f1_scores),3))
```
