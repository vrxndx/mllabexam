```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
data=pd.read_csv("/home/backup/ml_datasets/wine_quality_red.csv")
print(data)
print(data.info())
print(data.isnull().sum())
```

```python
x=data.drop("quality",axis=1)
y=data['quality']
print(x)
print(y)
```

```python
y=(y>=6).astype(int)
print("\nClass Distribution:")
print(y.value_counts())
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
print("Splitted")
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
models={
    "1 Hidden Layer - 10 Neurons":(10,),
    "1 Hidden Layer - 50 Neurons":(50,),
    "2 Hidden Layers - 50,25 Neurons ":(50,25),
    "2 Hidden Layers - 100,50 Neurons":(100,50),
    "3 Hidden Layers - 100,50,25 Neurons":(100,50,25),
    "3 Hidden Layers - 100,100,50 Neurons":(100,100,50),
}
results=[]
for name,architecture in models.items():
    print("\nTraining:",name)
    mlp=MLPClassifier(
        hidden_layer_sizes=architecture,
        activation='relu',
        solver='adam',
        max_iter=1000,
        random_state=42
    )
    mlp.fit(x_train_scaled,y_train)
    y_pred=mlp.predict(x_test_scaled)
    accuracy=accuracy_score(y_test,y_pred)
    precision=precision_score(y_test,y_pred)
    recall=recall_score(y_test,y_pred)
    f1=f1_score(y_test,y_pred)
    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1
    ])
    print("done")
        
    
```

```python
results_df=pd.DataFrame(
    results,
    columns=[
        "Architecture",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)
print("\n\nPerformance Comparison:")
print(results_df.to_string(index=False))
```

```python
#Identify best model
best_model=results_df.loc[
    results_df["F1 Score"].idxmax()
]
print("\nBest Architecture:")
print(best_model)
```
